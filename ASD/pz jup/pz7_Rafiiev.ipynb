**Умова:** Задано дві послідовності символів: $X = \text{"ABCDE"}$ і $Y = \text{"XYZABCD"}$. Необхідно знайти найдовшу спільну підпослідовність символів, використовуючи алгоритм Хаббарда.   
**Теоретична довідка про алгоритм Хаббарда**  
Алгоритм Хаббарда (Hirschberg's algorithm) — це оптимізація класичного динамічного програмування для пошуку LCS.  
 - Класичний метод потребує матриці розміром $O(m \cdot n)$ як для обчислення довжини, так і для відновлення рядка.
 - Алгоритм Хаббарда використовує підхід «розділяй і володарюй» (Divide and Conquer). Він обчислює лише поточний та попередній рядки матриці DP для пошуку точок розрізу, що знижує потребу в пам'яті до лінійної — $O(\min(m, n))$, зберігаючи часову складність $O(m \cdot n)$.  
**1. Програмна реалізація алгоритму Хаббарда на Python**  

```python
def lcs_length_row(x, y):
    m, n = len(x), len(y)
    curr = [0] * (n + 1)
    
    for i in range(1, m + 1):
        prev = list(curr)
        for j in range(1, n + 1):
            if x[i - 1] == y[j - 1]:
                curr[j] = prev[j - 1] + 1
            else:
                curr[j] = max(prev[j], curr[j - 1])
    return curr

def hirschberg(x, y):
    if len(x) == 0:
        return ""
    if len(x) == 1:
        return x if x in y else ""
        
    mid = len(x) // 2
    x_left = x[:mid]
    x_right = x[mid:]
    
    f = lcs_length_row(x_left, y)
    g = lcs_length_row(x_right[::-1], y[::-1])
    
    max_val = -1
    split_idx = 0
    n = len(y)
    for j in range(n + 1):
        if f[j] + g[n - j] > max_val:
            max_val = f[j] + g[n - j]
            split_idx = j

    return hirschberg(x_left, y[:split_idx]) + hirschberg(x_right, y[split_idx:])

s1 = "ABCDE"
s2 = "XYZABCD"

result_lcs = hirschberg(s1, s2)
print("Вхідний рядок 1:", s1)
print("Вхідний рядок 2:", s2)
print("Найдовша спільна підпослідовність (Алгоритм Гіршберга):", result_lcs)
print("Довжина LCS:", len(result_lcs))
```

3. Цей підхід ітеративно заповнює двовимірну таблицю $dp$ розміром $(m+1) \times (n+1)$, де кожна комірка $dp[i][j]$ обчислюється на основі попередніх станів: якщо символи рядків збігаються, значення збільшується на одиницю від діагонального сусіда зліва-зверху, а якщо ні — береться максимум із сусідніх комірок зверху або зліва, після чого сама підпослідовність відновлюється зворотним проходом від правої нижньої комірки до початку таблиці.
4. Алгоритм Хаббарда працює за принципом «розділяй і володарюй», розбиваючи перший рядок навпіл і знаходячи за допомогою лінійного масиву пам'яті оптимальну точку поділу для другого рядка (де сума довжин лівої частини та розгорнутої правої частини є максимальною), після чого рекурсивно запускає пошук для отриманих половин, що дозволяє відновити весь рядок без збереження повної двовимірної матриці.
5. Класичне динамічне програмування просте в реалізації та швидко виконує зворотний шлях відновлення рядка, проте його критичним недоліком є колосальне кубічне або квадратичне споживання пам'яті $O(m \cdot n)$, тоді як алгоритм Хаббарда має колосальну перевагу у вигляді лінійної складності пам'яті $O(\min(m, n))$, але його недоліком є складніша рекурсивна архітектура коду та незначне збільшення константи часу виконання через повторні обчислення масивів на кожному рівні поділу.
